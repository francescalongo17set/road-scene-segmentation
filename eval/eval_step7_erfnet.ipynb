```python
# Verify dataset structure and ERFNet weights
import os

BASE      = os.environ.get("MASKARCH_DATA_ROOT", "../data")  # see data/README.md
ANOMALY_DIR = BASE + "/datasets/anomaly"

print("=== Dataset structure ===")
for root, dirs, files in os.walk(ANOMALY_DIR):
    level = root.replace(ANOMALY_DIR, '').count(os.sep)
    print('  ' * level + os.path.basename(root) + '/')
    for f in sorted(files)[:3]:
        print('  ' * (level + 1) + f)

print("\n=== ERFNet weights ===")
weights_path = "../trained_models/erfnet_pretrained.pth"
print("Found:", os.path.exists(weights_path), "→", weights_path)
```
